```python
import spacy
nlp=spacy.load("en_core_web_lg")
doc=nlp("Hi Hellow throw put ")
for token in doc:
    print(token.text,"|",token.has_vector,"|",token.is_oov)
```

```python
com=nlp("hey")
```

```python
for token in doc:
    print(token.text,"|",token.similarity(com))
```
