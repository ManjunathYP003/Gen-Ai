```python
import spacy
nlp=spacy.load("en_core_web_sm")
```

```python
do=nlp("ElonMusk runs the Tesla")
```

```python
for token in do.ents:
    print(token.text,"|",token.label_)
```

```python
doc=nlp("Mr.ElonMusk runs the Tesla")
for token in doc.ents:
    print(token.text,"|",token.label_)
```

```python
from spacy.tokens import Span
s1=Span(do,0,1,label='PERSON')
do.set_ents([s1],default="unmodified")
```

```python
for token in do.ents:
    print(token.text,"|",token.label_)
```
