```python
import spacy
import pandas as pd
from spacy.lang.en.stop_words import STOP_WORDS
```

```python
data=pd.read_json("C:\\Users\\manju\\Downloads\\archive (30)\combined.json",lines=True)
```

```python
nlp=spacy.load("en_core_web_sm")
```

```python
def preprocess(doc):
    text=nlp(doc)
    not_space=[token.text for token in text if not token.is_stop and not token.is_punct]
    return " ".join(not_space)
```

```python
doc=("I am learning NLP and I am using spaCy for text processing.")
preprocess(doc)
```
