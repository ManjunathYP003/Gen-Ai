```python
from langchain_ollama import ChatOllama
llm=ChatOllama(model='granite4:350m',temperature=1.5,num_predict=20)
llm.invoke("The first person to go to moon is")
```

```python
from langchain_groq import ChatGroq
llm=ChatGroq(model="openai/gpt-oss-120b",temperature=1.5)
llm.invoke("The first person to go to moon is")
```

```python
from langchain_core.prompts import PromptTemplate
pt=PromptTemplate.from_template("write a brief note on {Topic}")
llm=ChatGroq(model="openai/gpt-oss-120b",temperature=1.5)
```

```python
chain=pt|llm
resp=chain.invoke({'Topic':'NLP'},{"Topic":"ML"})
```
