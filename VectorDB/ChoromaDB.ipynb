```python
import chromadb
client=chromadb.Client()
collection=client.create_collection(name="Data")
```

```python
collection.add(
    documents=[
        "This apple company is not releasing mobiles often",
        "Apple keeps you away from doctors "
    ],
    ids=['id1','id2']
)
```

```python
res=collection.query(
    query_texts=["New iphone will be expected  to be released in september"],
    n_results=2
)
```

```python
from datetime import datetime
client=chromadb.Client()
collection=client.create_collection(name="Rivews",
                                    metadata={
                                        'description':"Product_review",
                                        'created':str(datetime.now())
                                    })
```

```python
collection.add(
    documents=[ "This apple company is not releasing mobiles often",
            "Apple keeps you away from doctors ",
            "Doctor suggested to take atleast 2 times fruit diet in a week"],
            ids=["id1","id2",'id3'],
            metadatas=[
                {"Catgory":"Elctronics",'rating':8},
                {'Catgory':"Fruits",'rating':9},
                {'Catgory':"Fruits",'rating':9}
            ]
    
)
```

```python
collection.count()
```

```python
collection.peek()
```

```python
collection.query(query_texts=[
    "Maintainig a good diet is important to be fit"
],n_results=2)
```
