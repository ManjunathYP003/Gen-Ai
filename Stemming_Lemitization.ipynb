```python
import nltk
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
words=["eat",'eating','ate','abilty','adjustable','stability','Taken']
for word in words:
    print(word,'|',stemmer.stem(word))
```

```python
import spacy
nlp=spacy.load("en_core_web_sm")
data=nlp("eat eating ate ability adjustable highlight taken Brah Bro")
for token in data:
    print(token,"|",token.lemma_)
```

```python
nlp.pipe_names
```

```python
ar=nlp.get_pipe('attribute_ruler')
ar.add(
    patterns=[
        [{'TEXT':'Bro'}],
      [{"TEXT":'Brah'}]
      ],
      attrs={"LEMMA":"Brother"})
```

```python
doc=nlp("Bro is my friend. Brah helped me.")
for token in doc:
    print(token,"|",token.lemma_)
```

```python
doc=nlp("Bro is my   #friend. Brah helped me.")
for token in doc:
     if token.pos_ not in ["SPACE",'X',"PUNCT",'SYM']:
        print(token.text,"|",token.lemma_)
```
